## 1. Environment Setup & Dependencies
Import essential libraries: `transformers`, `pandas`, `sklearn`, `tabulate`, and `matplotlib`.

```python
import os
import pandas as pd
import numpy as np
from transformers import pipeline
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix
from tabulate import tabulate
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

## 2. Load Custom Dataset (20 Samples)
We load 20 realistic text samples representing real-world consumer reviews, support tickets, and service feedback.

```python
dataset_path = 'dataset.csv'
df = pd.read_csv(dataset_path)
print(f"Loaded {len(df)} samples.")
df.head()
```

## 3. Load Pretrained Sentiment Analysis Model
We initialize the Hugging Face `sentiment-analysis` pipeline backed by `distilbert-base-uncased-finetuned-sst-2-english`.

```python
model_name = 'distilbert-base-uncased-finetuned-sst-2-english'
classifier = pipeline('sentiment-analysis', model=model_name)
print(f"Pretrained model '{model_name}' initialized.")
```

## 4. Run Model Inference

```python
predictions = []
confidence_scores = []
correct_flags = []

for idx, row in df.iterrows():
    sample_text = row['text']
    ground_truth = row['ground_truth'].upper()
    
    res = classifier(sample_text)[0]
    pred_label = res['label'].upper()
    score = float(res['score'])
    
    predictions.append(pred_label)
    confidence_scores.append(round(score, 4))
    correct_flags.append('PASS' if pred_label == ground_truth else 'FAIL')

df['predicted'] = predictions
df['confidence'] = confidence_scores
df['status'] = correct_flags
df['correct_bool'] = (df['ground_truth'] == df['predicted'])

print("Inference completed for all 20 samples.")
```

## 5. Calculate Metrics & Confusion Matrix

```python
y_true = df['ground_truth']
y_pred = df['predicted']

acc = accuracy_score(y_true, y_pred)
precision, recall, f1, _ = precision_recall_fscore_support(y_true, y_pred, average='binary', pos_label='POSITIVE')
cm = confusion_matrix(y_true, y_pred, labels=['POSITIVE', 'NEGATIVE'])

print(f"Overall Accuracy     : {acc * 100:.2f}%")
print(f"Precision (POSITIVE) : {precision:.4f}")
print(f"Recall (POSITIVE)    : {recall:.4f}")
print(f"F1-Score (POSITIVE)  : {f1:.4f}")
```
